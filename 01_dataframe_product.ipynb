# DataFrame with Product Column

**Description:** Creates a DataFrame with 5 rows and 3 columns, then multiplies columns A and B into a new Product column.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

df['Product'] = df['A'] * df['B']
print(df)
```
