# Data Aggregation by Category

**Description:** Groups rows by Category and calculates the sum and mean of Value for each group.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': np.random.choice(['A', 'B', 'C'], size=10),
    'Value': np.random.randint(1, 100, size=10)
})

print("Original DataFrame:")
print(df)

result = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nSum and mean for each category:")
print(result)
```
