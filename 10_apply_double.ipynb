# Apply a Function to Double Values

**Description:** apply() applies the lambda function to each column; every value is multiplied by 2.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

print("Original DataFrame:")
print(df)

result = df.apply(lambda x: x * 2)

print("\nDataFrame after doubling:")
print(result)
```
