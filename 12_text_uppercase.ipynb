# Convert Text to Uppercase

**Description:** str.upper() converts every string in the Series to uppercase.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd

s = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])

print("Original Series:")
print(s)

print("\nUppercase Series:")
print(s.str.upper())
```
