# MultiIndex DataFrame: Indexing and Slicing

**Description:** Creates a two-level index. loc[] selects a category, a specific index pair, or a slice.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [('A', 1), ('A', 2), ('B', 1), ('B', 2)],
    names=['Category', 'Number']
)

df = pd.DataFrame({'Value': [10, 20, 30, 40]}, index=index)

print("MultiIndex DataFrame:")
print(df)

print("\nRows in Category A:")
print(df.loc['A'])

print("\nValue at ('B', 1):")
print(df.loc[('B', 1)])

print("\nSlice from ('A', 1) to ('B', 1):")
print(df.loc[('A', 1):('B', 1)])
```
