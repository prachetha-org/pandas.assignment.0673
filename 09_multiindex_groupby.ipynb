# MultiIndex with Category and SubCategory

**Description:** set_index() creates a hierarchical index. groupby() with both index levels calculates each pair's total.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Category': ['A', 'A', 'A', 'B', 'B', 'B'],
    'SubCategory': ['X', 'Y', 'X', 'X', 'Y', 'Y'],
    'Value': np.random.randint(1, 100, size=6)
})

df = df.set_index(['Category', 'SubCategory'])

print("MultiIndex DataFrame:")
print(df)

print("\nSum for each Category and SubCategory:")
print(df.groupby(level=['Category', 'SubCategory']).sum())
```
