# Row-wise and Column-wise Sum

**Description:** axis=1 calculates sums across each row. axis=0 calculates sums down each column.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(4, 3)),
                  columns=['A', 'B', 'C'])

print("DataFrame:")
print(df)

print("\nRow-wise sum:")
print(df.sum(axis=1))

print("\nColumn-wise sum:")
print(df.sum(axis=0))
```
