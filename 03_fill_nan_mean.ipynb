# Fill NaN Values with Column Mean

**Description:** Introduces missing values and replaces each NaN with the mean of its own column.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(5, 3)),
                  columns=['A', 'B', 'C'])

df.iloc[0, 1] = np.nan
df.iloc[2, 2] = np.nan

print("Before filling missing values:")
print(df)

df = df.fillna(df.mean())

print("\nAfter filling missing values with column means:")
print(df)
```
