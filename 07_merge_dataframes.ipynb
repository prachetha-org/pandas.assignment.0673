# Merge Two DataFrames

**Description:** pd.merge() combines two DataFrames using their common ID column.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ram', 'Sam', 'John']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [80, 90, 85]
})

result = pd.merge(df1, df2, on='ID')

print("Merged DataFrame:")
print(result)
```
