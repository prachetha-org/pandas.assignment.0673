# Extract First Three Characters

**Description:** str[:3] extracts characters at positions 0, 1, and 2 from each string.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd

s = pd.Series(['apple', 'banana', 'mango', 'orange', 'grapes'])

print("Original Series:")
print(s)

print("\nFirst three characters:")
print(s.str[:3])
```
