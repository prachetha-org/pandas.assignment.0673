# Total Sales by Category

**Description:** Groups products by Category and adds the Sales values in each category.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

products = ['Pen', 'Book', 'Pencil', 'Bag', 'Scale', 'Eraser']
categories = ['Stationery', 'Books', 'Stationery', 'Bags', 'Stationery', 'Stationery']

df = pd.DataFrame({
    'Product': products,
    'Category': categories,
    'Sales': np.random.randint(100, 1000, size=6)
})

print("Sales DataFrame:")
print(df)

result = df.groupby('Category')['Sales'].sum()
print("\nTotal sales by category:")
print(result)
```
