# Lambda Function to Create Row Sum

**Description:** Creates a Sum column by adding A, B, and C for each row. axis=1 applies the function row by row.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(6, 3)),
                  columns=['A', 'B', 'C'])

df['Sum'] = df.apply(lambda row: row['A'] + row['B'] + row['C'], axis=1)

print(df)
```
