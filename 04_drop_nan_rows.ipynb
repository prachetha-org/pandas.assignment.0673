# Drop Rows Containing NaN Values

**Description:** dropna() removes every row that contains at least one missing value.

**Libraries:** Pandas and NumPy (where required).

```python
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 10, size=(6, 4)),
                  columns=['A', 'B', 'C', 'D'])

df.iloc[1, 2] = np.nan
df.iloc[3, 1] = np.nan

print("Before dropna():")
print(df)

df = df.dropna()

print("\nAfter dropping rows containing any NaN:")
print(df)
```
